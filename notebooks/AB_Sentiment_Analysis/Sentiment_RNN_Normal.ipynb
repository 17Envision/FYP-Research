```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize

# Load stopwords from a file
with open('../../data/mnt/stopwords.txt', 'r', encoding='utf-8') as file:
    sinhala_stopwords = file.read().splitlines()

# Function to preprocess text
def preprocess(text):
    if isinstance(text, float):
        text = ''  # Convert NaN or float entries to empty string
    tokens = word_tokenize(text)
    filtered_tokens = [word for word in tokens if word not in sinhala_stopwords]
    return ' '.join(filtered_tokens)

# Load the dataset
file_path = '../../data/mnt/sentiment_Aspect_added.xlsx'
df = pd.read_excel(file_path)

# Apply preprocessing
df['Cleaned_Comment'] = df['Comment'].apply(preprocess)

# Display the first few rows to verify
print(df.head())
```

text preprocessing

Convert english into simple

```python
import re

# Define a function for selective text preprocessing
def preprocess_text(text):
    if isinstance(text, str):
        # Convert only English capital letters to lowercase
        return re.sub(r'[A-Z]', lambda match: match.group(0).lower(), text)
    return text

# Apply the preprocessing function to the 'Comment' column in the existing DataFrame 'data'
data['Comment'] = data['Comment'].apply(preprocess_text)

data.head(10)
```

remove links

```python
def remove_links(text):
    if isinstance(text, str):
        # Remove URLs starting with http or https
        text = re.sub(r'http\S+', '', text)
        # Remove URLs starting with www
        text = re.sub(r'www\S+', '', text)
        return text
        
# Apply the function to the 'Comment' column
df['Comment'] = df['Comment'].apply(remove_links)
```

Remove Punctuations

```python
import string
def remove_punctuation(text):
    if isinstance(text, str):
        # Remove punctuation marks
        text = text.translate(str.maketrans('', '', string.punctuation))
    return text

# Apply the function to the 'Comment' column
data['Comment'] = data['Comment'].apply(remove_punctuation)
```

Vectorization

```python
def vectorizer(ds, vocabulary):
    vectorized_lst = []
    
    for sentence in ds:
        sentence_lst = np.zeros(len(vocabulary))
        
        for i in range(len(vocabulary)):
            if vocabulary[i] in sentence.split():
                sentence_lst[i] = 1
                
        vectorized_lst.append(sentence_lst)
        
    vectorized_lst_new = np.asarray(vectorized_lst, dtype=np.float32)
    
    return vectorized_lst_new
```
